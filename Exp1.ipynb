```python
# Import Pandas library
import pandas as pd

# Read the CSV file
df = pd.read_csv("/content/drive/MyDrive/ML/car data.csv")

# Display complete dataset
print("Complete Dataset:")
print(df)

# Display first 5 records
print("\nFirst 5 Records:")
print(df.head())

# Display last 5 records
print("\nLast 5 Records:")
print(df.tail())

# Display shape
print("\nShape of Dataset:")
print(df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Select specific columns
print("\nSelected Columns:")
print(df[["Car_Name", "Year", "Selling_Price", "Fuel_Type"]])

# Calculate average selling price
print("\nAverage Selling Price:")
print(df["Selling_Price"].mean())

# Find maximum selling price
print("\nMaximum Selling Price:")
print(df["Selling_Price"].max())

# Find minimum selling price
print("\nMinimum Selling Price:")
print(df["Selling_Price"].min())
```
