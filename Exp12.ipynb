```python
# Experiment 12 - Association Rule Mining using Apriori Algorithm

# ---------------------------------------
# 1. Import Libraries
# ---------------------------------------

import pandas as pd
import matplotlib.pyplot as plt

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules


# ---------------------------------------
# 2. Load Dataset
# ---------------------------------------

df = pd.read_csv('/content/drive/MyDrive/ML/Groceries_dataset.csv')

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns)


# ---------------------------------------
# 3. Create Transactions
# ---------------------------------------

transactions = df.groupby(
    ['Member_number', 'Date']
)['itemDescription'].apply(list).tolist()

print("\nFirst 5 Transactions:")
for transaction in transactions[:5]:
    print(transaction)


# ---------------------------------------
# 4. One-Hot Encoding
# ---------------------------------------

te = TransactionEncoder()

encoded_data = te.fit(transactions).transform(transactions)

basket = pd.DataFrame(
    encoded_data,
    columns=te.columns_
)

print("\nOne-Hot Encoded Data:")
print(basket.head())


# ---------------------------------------
# 5. Generate Frequent Itemsets
# ---------------------------------------

frequent_itemsets = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)

frequent_itemsets = frequent_itemsets.sort_values(
    by='support',
    ascending=False
)

print("\nFrequent Itemsets:")
print(frequent_itemsets.head(20))


# ---------------------------------------
# 6. Generate Association Rules
# ---------------------------------------

rules = association_rules(
    frequent_itemsets,
    metric='confidence',
    min_threshold=0.1
)


# ---------------------------------------
# 7. Filter Rules Using Lift
# ---------------------------------------

rules = rules[rules['lift'] > 1]

rules = rules.sort_values(
    by='lift',
    ascending=False
)

print("\nAssociation Rules:")

print(
    rules[
        [
            'antecedents',
            'consequents',
            'support',
            'confidence',
            'lift'
        ]
    ].head(20)
)


# ---------------------------------------
# 8. Display Rules in Simple Form
# ---------------------------------------

print("\nTop Rules:")

for _, row in rules.head(10).iterrows():

    print(
        list(row['antecedents']),
        "->",
        list(row['consequents']),
        "| Support:",
        round(row['support'], 3),
        "| Confidence:",
        round(row['confidence'], 3),
        "| Lift:",
        round(row['lift'], 3)
    )


# ---------------------------------------
# 9. Top Frequent Items Graph
# ---------------------------------------

item_support = basket.mean().sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10, 5))

item_support.plot(kind='bar')

plt.xlabel("Items")
plt.ylabel("Support")
plt.title("Top 10 Frequent Grocery Items")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
