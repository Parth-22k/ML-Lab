```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder

# Load dataset
df = pd.read_csv("/content/drive/MyDrive/ML/car data.csv")

print("Original Data:")
print(df.head())


# -----------------------------
# 1. LABEL ENCODING
# -----------------------------

label_encoder = LabelEncoder()

df["Car_Name"] = label_encoder.fit_transform(df["Car_Name"])

print("\nAfter Label Encoding:")
print(df.head())


# -----------------------------
# 2. ONE-HOT ENCODING
# -----------------------------

one_hot_encoder = OneHotEncoder(sparse_output=False)

encoded_data = one_hot_encoder.fit_transform(
    df[["Fuel_Type", "Selling_type", "Transmission"]]
)

# Get new column names
new_columns = one_hot_encoder.get_feature_names_out(
    ["Fuel_Type", "Selling_type", "Transmission"]
)

# Convert encoded data into DataFrame
encoded_df = pd.DataFrame(
    encoded_data,
    columns=new_columns
)

# Remove original categorical columns
df = df.drop(
    ["Fuel_Type", "Selling_type", "Transmission"],
    axis=1
)

# Add encoded columns
df = pd.concat([df, encoded_df], axis=1)

print("\nAfter One-Hot Encoding:")
print(df.head())
```
